```python
from paddleocr import PPStructureV3


# Initialize the pipeline explicitly calling the Hugging Face transformers engine
pipeline = PPStructureV3(
    lang="eng",
    device="gpu",
    text_detection_model_name="PP-OCRv6_medium_det",
    text_recognition_model_name="PP-OCRv6_medium_rec",

    # Document optimization parameters
    use_doc_orientation_classify=True,
    use_doc_unwarping=True,

    # Advanced modular extraction
    use_formula_recognition=True,
    use_table_recognition=True,
    use_chart_recognition=True,
    use_seal_recognition=True,
    use_region_detection=True,
    use_textline_orientation=True
)
```

```python
# Run the extraction (using the correct input keyword)
results = pipeline.predict(
    input="document.png",
)

for res in results:
    res.save_to_markdown(save_path="output")
    res.save_to_json(save_path="output")

print("Successfully parsed multi-column document with the Transformers Engine!")
```
